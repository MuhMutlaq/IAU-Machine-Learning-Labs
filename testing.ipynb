```python
import model_training as training
import pandas as pd
import joblib

# 1. Load the model (if you are in a new session)
# If you just ran the training code, "model" is already in memory, so this is optional.
model= joblib.load("trained_model3.pkl")

# 2. Select a sample of data to test
sample_data= training.X_test.head()
sample_actual_prices= training.y_test.head()

# 3. Predict the prices for this sample
sample_predictions= model.predict(sample_data)

# 4. Create a DataFrame to compare Actual vs. Predicted
comparison_df= pd.DataFrame({
    "Actual Price": sample_actual_prices,
    "Predicted Price": sample_predictions,
    "Difference (Error)": sample_actual_prices - sample_predictions
})

pd.options.display.float_format= '{:,.2f}'.format

# 5. Display the comparison
print("---- Prediction Results for First 5 Test Houses ----")
display(comparison_df)

# 6. Predict for a specific single house
single_house= training.X_test.iloc[[0]] # Double brackets keeps it as a DataFrame
single_prediction= model.predict(single_house)
```
