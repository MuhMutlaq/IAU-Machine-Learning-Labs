## kdeplot

kdeplots are [Kernel Density Estimation plots](http://en.wikipedia.org/wiki/Kernel_density_estimation#Practical_estimation_of_the_bandwidth). These KDE plots replace every single observation with a Gaussian (Normal) distribution centered around that value. For example:

```python
# Don't worry about understanding this code!
# It's just for the diagram below
import numpy as np # Import numpy shortened as np
import matplotlib.pyplot as plt # Import specific library shortened as plt
from scipy import stats # Import stats from scipy

#Create dataset
dataset = np.random.randn(25) # Random numbers

# Create another rugplot
sns.rugplot(dataset); # Use rug plot

# Set up the x-axis for the plot
x_min = dataset.min() - 2 # The minimum value
x_max = dataset.max() + 2 # The maximum value

# 100 equally spaced points from x_min to x_max
x_axis = np.linspace(x_min,x_max,100) # The number of spaced numbers

# Set up the bandwidth, for info on this:
url = 'http://en.wikipedia.org/wiki/Kernel_density_estimation#Practical_estimation_of_the_bandwidth' # URL of the density

bandwidth = ((4*dataset.std()**5)/(3*len(dataset)))**.2 # The bandwidth of the dataset


# Create an empty kernel list
kernel_list = [] # List

# Plot each basis function
for data_point in dataset: # Loop that use the dataset
    
    # Create a kernel for each point and append to list
    kernel = stats.norm(data_point,bandwidth).pdf(x_axis) # Normalize the data from the bandwidth of the x axis spaced numbers (specified before)
    kernel_list.append(kernel) # List append
    
    #Scale for plotting
    kernel = kernel / kernel.max() # Scaling the kernel
    kernel = kernel * .4 # First 4 digits
    plt.plot(x_axis,kernel,color = 'grey',alpha=0.5) # Plot in green color that is 0.5 transparency for the figure

plt.ylim(0,1) # The limit of the figure is between 0 and 1
```

```python
# To get the kde plot we can sum these basis functions.

# Plot the sum of the basis function
sum_of_kde = np.sum(kernel_list,axis=0) # Sum of kernel list

# Plot figure
fig = plt.plot(x_axis,sum_of_kde,color='indianred') # Plot in specific color

# Add the initial rugplot
sns.rugplot(dataset,c = 'indianred') # Use rug plot

# Get rid of y-tick marks
plt.yticks([]) # Get rid of y ticks in y axis in the figure

# Set title
plt.suptitle("Sum of the Basis Functions") # The title of the figure 
```

# Categorical Data Plots

Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

* factorplot
* boxplot
* violinplot
* stripplot
* swarmplot
* barplot
* countplot

Let's go through examples of each!

```python
import seaborn as sns # Import seaborn shortened as sns
%matplotlib inline # Display the figures for each plot
```

```python
tips = sns.load_dataset('tips') # Load the seaborn dataset downloaded on it
tips.head()
```

## barplot and countplot

These very similar plots allow you to get aggregate data off a categorical feature in your data. **barplot** is a general plot that allows you to aggregate the categorical data based off some function, by default the mean:

```python
sns.barplot(x='sex',y='total_bill',data=tips) # Use bar plot
```

```python
import numpy as np # Import numpy shortened as np
```

You can change the estimator object to your own function, that converts a vector to a scalar:

```python
sns.barplot(x='sex',y='total_bill',data=tips,estimator=np.std) # Use bar plot that use std estimator
```

### countplot

This is essentially the same as barplot except the estimator is explicitly counting the number of occurrences. Which is why we only pass the x value:

```python
sns.countplot(x='sex',data=tips) # Use count plot
```

## boxplot and violinplot

boxplots and violinplots are used to shown the distribution of categorical data. A box plot (or box-and-whisker plot) shows the distribution of quantitative data in a way that facilitates comparisons between variables or across levels of a categorical variable. The box shows the quartiles of the dataset while the whiskers extend to show the rest of the distribution, except for points that are determined to be “outliers” using a method that is a function of the inter-quartile range.

```python
sns.boxplot(x="day", y="total_bill", data=tips,palette='rainbow') # Use box plot for specific color in the figure
```

```python
# Can do entire dataframe with orient='h'
sns.boxplot(data=tips,palette='rainbow',orient='h') # Use box plot for specific color and do an entire dataframe in the figure
```

```python
sns.boxplot(x="day", y="total_bill", hue="smoker",data=tips, palette="coolwarm") # Use box plot for specific color and hue in the figure
```

### violinplot
A violin plot plays a similar role as a box and whisker plot. It shows the distribution of quantitative data across several levels of one (or more) categorical variables such that those distributions can be compared. Unlike a box plot, in which all of the plot components correspond to actual datapoints, the violin plot features a kernel density estimation of the underlying distribution.

```python
sns.violinplot(x="day", y="total_bill", data=tips,palette='rainbow') # Use violin plot
```

```python
sns.violinplot(x="day", y="total_bill", data=tips,hue='sex',palette='Set1') # Use violin plot
```

```python
sns.violinplot(x="day", y="total_bill", data=tips,hue='sex',split=True,palette='Set1') # Use violin plot
```

## stripplot and swarmplot
The stripplot will draw a scatterplot where one variable is categorical. A strip plot can be drawn on its own, but it is also a good complement to a box or violin plot in cases where you want to show all observations along with some representation of the underlying distribution.

The swarmplot is similar to stripplot(), but the points are adjusted (only along the categorical axis) so that they don’t overlap. This gives a better representation of the distribution of values, although it does not scale as well to large numbers of observations (both in terms of the ability to show all the points and in terms of the computation needed to arrange them).

```python
sns.stripplot(x="day", y="total_bill", data=tips, palette='rainbow') # Use strip plot
```

```python
sns.stripplot(x="day", y="total_bill", data=tips,jitter=True, palette='rainbow') # Use strip plot and jitter type
```

```python
sns.stripplot(x="day", y="total_bill", data=tips,jitter=True,hue='sex',palette='Set1') # Use strip plot hue in the figure
```

```python
sns.stripplot(x="day", y="total_bill", data=tips,jitter=True,hue='sex',palette='Set1',dodge=True) # Use strip plot and dodge in the figure
```

```python
sns.swarmplot(x="day", y="total_bill", data=tips) # Use swarm plot
```

### Combining Categorical Plots

```python
sns.violinplot(x="tip", y="day", data=tips,palette='rainbow') # Use violin plot
sns.swarmplot(x="tip", y="day", data=tips,color='black',size=3) # Use swarm plot in specific size for the figure
```

## catplot

factorplot is the most general form of a categorical plot. It can take in a kind parameter to adjust the plot type:

```python
sns.catplot(x='sex',y='total_bill',data=tips,kind='bar') # Use cat plot
```

## Heatmap

In order for a heatmap to work properly, your data should already be in a matrix form, the sns.heatmap function basically just colors it in for you. For example:

```python
tips.head() # Display first 5 rows
```

```python
# Matrix form for correlation data
tips.corr() # Display the matrix form for correlation of the tips info
```

```python
sns.heatmap(tips.corr()) # Display the heatmap of correlation data of the tips info
```

```python
sns.heatmap(tips.corr(),cmap='coolwarm',annot=True) # Draw the figure using heatmap in corr() data of the tips info
```

Or for the flights data:

```python
flights.pivot_table(values='passengers',index='month',columns='year') # Draw the pivot table
```

```python
pvflights = flights.pivot_table(values='passengers',index='month',columns='year') # Draw the pivot table
sns.heatmap(pvflights) # Draw the figure using heatmap
```

```python
sns.heatmap(pvflights,cmap='magma',linecolor='white',linewidths=1) # Draw the figure using heatmap from specified wdith, color, and cmap for the flights info
```

## clustermap

The clustermap uses hierarchal clustering to produce a clustered version of the heatmap. For example:

```python
sns.clustermap(pvflights) # Draw the cluster map
```

Notice now how the years and months are no longer in order, instead they are grouped by similarity in value (passenger count). That means we can begin to infer things from this plot, such as August and July being similar (makes sense, since they are both summer travel months)

```python
# More options to get the information a little clearer like normalization
sns.clustermap(pvflights,standard_scale=1) # Draw the cluster map from specified standard_scale= 1
```

# Regression Plots

Seaborn has many built-in capabilities for regression plots, however we won't really discuss regression until the machine learning section of the course, so we will only cover the **lmplot()** function for now.

**lmplot** allows you to display linear models, but it also conveniently allows you to split up those plots based off of features, as well as coloring the hue based off of features.

Let's explore how this works:

```python
import seaborn as sns # Shortened import of seaborn as sns
%matplotlib inline # Display the figures for each plot
```

```python
tips = sns.load_dataset('tips') # Load embedded dataset from seaborn
```

```python
tips.head() # Display first 5 rows
```

## lmplot()

```python
sns.lmplot(x='total_bill',y='tip',data=tips) # Draw the figure using lmplot type in specified data
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex') # Draw the figure using lmplot type in specified data
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex',palette='coolwarm') # Draw the figure using lmplot type in specified data
```

## Using a Grid

We can add more variable separation through columns and rows with the use of a grid. Just indicate this with the col or row arguments:

```python
sns.lmplot(x='total_bill',y='tip',data=tips,col='sex') # Draw the figure using lmplot type in specified data
```

```python
sns.lmplot(x="total_bill", y="tip", row="sex", col="time",data=tips) # Draw the figure using lmplot type in specified data
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,col='day',hue='sex',palette='coolwarm') # Draw the figure using lmplot type in specified data
```

## Aspect and Size

Seaborn figures can have their size and aspect ratio adjusted with the **height** and **aspect** parameters:

```python
sns.lmplot(x='total_bill',y='tip',data=tips,col='day',hue='sex',palette='coolwarm',
          aspect=0.6,height=8) # Draw the figure using lmplot type in specified data
```
